```python
from transformers import BertTokenizer, BertModel
import torch

tokenizer = BertTokenizer.from_pretrained("google-bert/bert-base-chinese")
model = BertModel.from_pretrained("google-bert/bert-base-chinese")

# Function to generate BERT embeddings for a poem
def get_poem_embedding(poem, tokenizer, model):
    inputs = tokenizer(poem, return_tensors="pt", padding=True, truncation=True)
    outputs = model(**inputs)
    # Average embeddings to get a single vector for the poem
    poem_embedding = outputs.last_hidden_state.mean(dim=1)
    return poem_embedding

# Define two poems
poem1 = "城市夜景灯火明，水中倒影似星辰。高楼林立如画卷，繁华盛世醉人心。"
poem2 = "庭院深深绿意浓，盆栽满布似春风。翠叶摇曳随时舞，花香阵阵入心中。"

# Get embeddings for both poems
embedding1 = get_poem_embedding(poem1, tokenizer, model)
embedding2 = get_poem_embedding(poem2, tokenizer, model)

# Add the two embeddings to get a combined vector
combined_embedding = embedding1 + embedding2
print(combined_embedding)
```

```python
import torch
import numpy as np
```

```python
embedding_vector = combined_embedding.clone().detach()

vocab = tokenizer.get_vocab()
word_embeddings = model.embeddings.word_embeddings.weight.clone().detach()
valid_indices = [idx for token, idx in vocab.items() if not token.startswith("##") and token != "[MASK]"]

# Calculate cosine similarity only on valid indices
filtered_embeddings = word_embeddings[valid_indices]
similarities = torch.nn.functional.cosine_similarity(embedding_vector, filtered_embeddings, dim=-1)
closest_word_index = similarities.argmax().item()

# Find the corresponding token for the closest word
closest_word_token = list(vocab.keys())[valid_indices[closest_word_index]]
print(f"The closest word is: {closest_word_token}")
```
