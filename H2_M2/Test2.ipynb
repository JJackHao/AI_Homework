```python
import torch
from transformers import BertTokenizer, BertModel

# Load the model and tokenizer
tokenizer = BertTokenizer.from_pretrained("google-bert/bert-base-chinese")
model = BertModel.from_pretrained("google-bert/bert-base-chinese")
```

```python
# Encode each poem
poem1 = "城市夜景灯火明，水中倒影似星辰。高楼林立如画卷，繁华盛世醉人心。"  # First poem text
poem2 = "庭院深深绿意浓，盆栽满布似春风。翠叶摇曳随时舞，花香阵阵入心中。"  # Second poem text

tokens1 = tokenizer(poem1, return_tensors="pt")
tokens2 = tokenizer(poem2, return_tensors="pt")

with torch.no_grad():
    embedding1 = model(**tokens1).last_hidden_state.mean(dim=1)
    embedding2 = model(**tokens2).last_hidden_state.mean(dim=1)


# Simple Average
combined_average = (embedding1 + embedding2) / 2

# Weighted Average
w1, w2 = 0.7, 0.3
combined_weighted = w1 * embedding1 + w2 * embedding2

# Element-wise Multiplication
combined_hadamard = embedding1 * embedding2

# Geometric Mean
combined_geo_mean = (embedding1 * embedding2).sqrt()

# Difference Vector
combined_difference = embedding1 - embedding2

# Maximum Pooling
combined_max_pool = torch.max(embedding1, embedding2)


# To convert combined vector to a poem, use nearest neighbor approach or generation model
# Example: Get closest words to approximate poem creation (or use a language model to generate directly)

# Finding closest words (simplified, may require looping to create a full poem)
vocab = tokenizer.get_vocab()
word_embeddings = model.embeddings.word_embeddings.weight
similarities = torch.nn.functional.cosine_similarity(combined_weighted, word_embeddings, dim=-1)

# Get top N words closest to combined embedding
top_k = 28  # Define how many words you want to sample
top_k_indices = similarities.topk(top_k).indices
new_poem_tokens = [tokenizer.convert_ids_to_tokens(idx.item()) for idx in top_k_indices]
new_poem = " ".join(new_poem_tokens)
print(f"Generated Poem: {new_poem}")
```
