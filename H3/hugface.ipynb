```python
# Use a pipeline as a high-level helper
from transformers import pipeline
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()
```

```python
def img2text(imgUrl):
    img_to_text_pipe = pipeline("image-to-text", model="Salesforce/blip-image-captioning-large")
    text = img_to_text_pipe(imgUrl)[0]["generated_text"]
    return text
```

```python
scenario = img2text("2.jpg")
print(scenario)
```

```python
def poemGeneration(msg):
  client = OpenAI()
  msg_list = [{"role": "system", "content": "You are a talented poet, expanding a short sentence into a Chinese poem with English translation."}]
  msg_list.append(msg)

  completion = client.chat.completions.create(
    model="gpt-4o",
    temperature=0.2,
    max_completion_tokens=200,
    messages= msg_list
  )
  out_msg = completion.choices[0].message.content
  return out_msg
```

```python
msg = {"role": "user", "content": scenario}
poem = poemGeneration(msg)
print(poem)
```
